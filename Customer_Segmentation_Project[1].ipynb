# Customer Segmentation Project
### Segment customers based on behavior and demographics

**Objective:** Use K-Means clustering to identify meaningful customer groups and derive targeted marketing insights.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

df = pd.read_csv('customer_segmentation_data.csv')
df.head()
```

## 1. Data Overview

```python
print(df.shape)
display(df.describe(include='all'))
print('\nMissing values:')
print(df.isnull().sum())
```

## 2. Exploratory Data Analysis

```python
plt.figure(figsize=(9,6))
plt.scatter(df['AnnualIncome_k'], df['SpendingScore'], alpha=0.6)
plt.xlabel('Annual Income')
plt.ylabel('Spending Score')
plt.title('Income vs Spending Score')
plt.grid(alpha=0.2)
plt.show()
```

## 3. Feature Scaling

```python
features = ['Age','AnnualIncome_k','PurchaseFrequency','AverageOrderValue',
            'WebsiteVisitsPerMonth','DiscountUsagePct','SpendingScore']
scaler = StandardScaler()
X = scaler.fit_transform(df[features])
```

## 4. Find the Number of Clusters

```python
inertias, silhouettes = [], []
for k in range(2,9):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = model.fit_predict(X)
    inertias.append(model.inertia_)
    silhouettes.append(silhouette_score(X, labels))

plt.figure(figsize=(9,5))
plt.plot(range(2,9), inertias, marker='o')
plt.title('Elbow Method')
plt.xlabel('k')
plt.ylabel('Inertia')
plt.show()

plt.figure(figsize=(9,5))
plt.plot(range(2,9), silhouettes, marker='o')
plt.title('Silhouette Scores')
plt.xlabel('k')
plt.ylabel('Silhouette Score')
plt.show()
```

## 5. K-Means Clustering

The project uses **4 clusters** for the final segmentation model.

```python
best_k = 4
kmeans = KMeans(n_clusters=best_k, random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(X)
df.groupby('Cluster')[features].mean().round(2)
```

## 6. Visualize the Customer Segments

```python
plt.figure(figsize=(10,7))
for c in sorted(df['Cluster'].unique()):
    subset = df[df['Cluster'] == c]
    plt.scatter(subset['AnnualIncome_k'], subset['SpendingScore'],
                s=30, alpha=0.65, label=f'Cluster {c}')
plt.xlabel('Annual Income')
plt.ylabel('Spending Score')
plt.title('Customer Segmentation')
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```

## 7. Business Insights
- **High-value customers:** prioritize loyalty rewards, premium products and personalized offers.
- **Deal seekers:** use coupons, bundles and limited-time discounts.
- **Regular customers:** encourage repeat purchases with loyalty programs and cross-selling.
- **Low-value/occasional customers:** use re-engagement campaigns and personalized recommendations.
